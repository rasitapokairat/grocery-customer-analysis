```python
# Import related libraries

import os
import numpy
import pandas
import seaborn
import geopandas
import plotly.express
import matplotlib.pyplot

# Disable warning
import warnings
warnings.filterwarnings("ignore")
```

```python
# TESCO 1.0 DATASET

# Declare vars
folder_path = "/Users/rasitapkr/Documents/MSc Data Science/Applied Data Science/Coursework Sem 2/Datasets/"
folder_path_groceriespurchased = folder_path + "Area-level grocery purchases/"

# Get annual data from file and add to dataframe

annual_data_groceriespurchased_borough = pandas.DataFrame([])
annual_data_groceriespurchased_borough = pandas.read_csv(folder_path_groceriespurchased + "year_borough_grocery.csv", skipinitialspace = True)
display(annual_data_groceriespurchased_borough)
```

```python
# Prepare data to extract INSIGHT 1 from TESCO 1.0 dataset

# # Disable warning
# pandas.set_option("mode.chained_assignment", None)

# Get data
area_population_borough = annual_data_groceriespurchased_borough.copy()

# Calculate gender different ratio for plotting graph
area_population_borough["gender_majority"] = numpy.where(area_population_borough["male"] > area_population_borough["female"], "male", \
                                                         numpy.where(area_population_borough["male"] < area_population_borough["female"], "female", "equal"))
area_population_borough["male_percentage"] = (area_population_borough["male"] * 100) / area_population_borough["population"]
area_population_borough["female_percentage"] = (area_population_borough["female"] * 100) / area_population_borough["population"]
area_population_borough["gender_diff"] = area_population_borough["male_percentage"] - area_population_borough["female_percentage"]

# Remove unused columns
insight_1_cols = ["area_id", "f_beer", "f_dairy", "f_eggs", "f_fats_oils", "f_fish", "f_fruit_veg", "f_grains", "f_meat_red", "f_poultry", \
                  "f_readymade", "f_sauces", "f_soft_drinks", "f_spirits", "f_sweets", "f_tea_coffee", "f_water", "f_wine", \
                  "gender_majority", "male_percentage", "female_percentage", "gender_diff"]
area_population_borough.drop(area_population_borough.columns.difference(insight_1_cols), axis = 1, inplace = True)

# Select male top 5 and female top 5 areas
area_population_borough_maletop5 = area_population_borough.sort_values("male_percentage",ascending = False).head(5)
area_population_borough_femaletop5 = area_population_borough.sort_values("female_percentage",ascending = False).head(5)
area_population_borough_bothtop5 = area_population_borough_maletop5.append(area_population_borough_femaletop5, ignore_index=True)

# Display data
display(area_population_borough_bothtop5)
area_population_borough_bothtop5.to_csv(folder_path + "year_borough_grocery_addedpop.csv", index = False, header = True)


```

```python
# Prepare data to extract INSIGHT 2 from TESCO 1.0 dataset

# Get data
fruitnveg_borough = annual_data_groceriespurchased_borough.copy()

# Remove unused columns
insight_2_cols = ["area_id", "f_beer", "f_dairy", "f_eggs", "f_fats_oils", "f_fish", "f_fruit_veg", "f_grains", "f_meat_red", "f_poultry", \
                  "f_readymade", "f_sauces", "f_soft_drinks", "f_spirits", "f_sweets", "f_tea_coffee", "f_water", "f_wine"]
fruitnveg_borough.drop(fruitnveg_borough.columns.difference(insight_2_cols), axis = 1, inplace = True)

# Select top 5 and bottom 5 areas
fruitnveg_borough_top5 = fruitnveg_borough.sort_values("f_fruit_veg",ascending = False).head(5)
fruitnveg_borough_bottom5 = fruitnveg_borough.sort_values("f_fruit_veg",ascending = True).head(5)
fruitnveg_borough = fruitnveg_borough_top5.append(fruitnveg_borough_bottom5, ignore_index=True)
fruitnveg_borough = fruitnveg_borough.sort_values("f_fruit_veg",ascending = False, ignore_index=True)

# Display data
display(fruitnveg_borough)
fruitnveg_borough.to_csv(folder_path + "year_borough_grocery_fruitnveg.csv", index = False, header = True)
```

```python
# REGIONAL GROSS DISPOSABLE HOUSEHOLD INCOME DATASET

# Declare vars
folder_path = "/Users/rasitapkr/Documents/MSc Data Science/Applied Data Science/Coursework Sem 2/Datasets/"

# Get annual data from file and add to dataframe
income_borough = pandas.DataFrame([])
income_borough_file = pandas.ExcelFile(folder_path + "regionalgrossdisposablehouseholdincomelocalauthoritytlilondon2020.xls")
income_borough = pandas.read_excel(income_borough_file, "Table 2", skiprows=[0])
income_borough = income_borough.sort_values("LAD code",ascending = True, ignore_index=True)
display(income_borough)
```

```python
# Prepare data to extract INSIGHT from both datasets

# Get income 2015 data
income_borough_2015 = income_borough.copy()
insight_both_cols = ["LAD code", "2015"]
income_borough_2015.drop(income_borough_2015.columns.difference(insight_both_cols), axis = 1, inplace = True)

# Convert income to represent in 100K
income_borough_2015["2015 in 100K"] = income_borough_2015["2015"] / 100000
# display(income_borough_2015)

# Combine 2 datasets together
two_datasets = annual_data_groceriespurchased_borough.merge(income_borough_2015, left_on="area_id", right_on="LAD code")
insight_both_cols = ["area_id", "f_fruit_veg", "2015 in 100K"]
two_datasets.drop(two_datasets.columns.difference(insight_both_cols), axis = 1, inplace = True)
display(two_datasets)

# Export
two_datasets.to_csv(folder_path + "year_borough_grocery_withincome.csv", index = False, header = True)
```
